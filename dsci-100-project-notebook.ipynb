#### Task 3.1: Tidy the data (5 points)

Produce a tidy data set that could be used to answer your research question using the `tidyverse`. Please ensure you have at least one quantitative (numeric) and one categorical variable:
- Create new variables (if needed)
- Transform the data into a tidy form (if needed)

- Comment your code throughout

```r
library(tidyverse)
WMM_tidy_data <- WMM_data  |> 
                    select(year, school, seed, conf_wins_pct, reg_wins_pct, tourney_finish)  |> # first, I use select to select for 6 the columns that I need
                    filter(year >= 1986)  |> # the Womens March Madness only added second_round_loss as a tourney_finish outcome in 1986 (the year they expanded to 40 teams). To keep the tourney_finish outcomes consistent, I only used data from 1986 and onwards.
                    mutate(tourney_finish = as_factor(tourney_finish)) |> # the tourney_finish data was previously stored as characters. It's better to convert it to a factor data type because it is a categorical variable.
                    mutate(tourney_finish = fct_recode(tourney_finish, "srl" = "second_round_loss", "frl" = "first_round_loss", "t16" = "top_16_loss" ,"t8" = "top_8_loss", "t4" = "top_4_loss", "t2" = "top_2_loss")) |> # I used fct_recode to abbreviate the factor value names for the variable Im trying to predict to simplify future analysis.
                    drop_na() # I want to remove all NAs from the data in case they mess up the analysis
head(WMM_tidy_data)
```

#### Task 4.1: Calculate summary statistics of numeric variables (2 points)

- Use any functions from the `tidyverse` to create **one** table that outputs the mean, minimum, and maximum of **all** numeric columns in your data, dropping the missing values if they exist.
- Show the outputted table.

</div>

```r
Year_summary <- WMM_tidy_data |> # First, I created summary statistics for each variable (Year, Seed, Conf_win_pct, and Reg_wins_Pct) using summarize(), and I dropped all missing values using na.rm = TRUE
                                summarize(min = min(year, na.rm = TRUE),
                                          max = max(year, na.rm = TRUE),
                                          mean = mean(year, na.rm = TRUE)
                                          )
Seed_summary <- WMM_tidy_data |> 
                                summarize(min = min(seed, na.rm = TRUE),
                                          max = max(seed, na.rm = TRUE),
                                          mean = mean(seed, na.rm = TRUE)
                                          )
conf_wins_pct_summary <- WMM_tidy_data |> 
                                summarize(min = min(conf_wins_pct, na.rm = TRUE),
                                          max = max(conf_wins_pct, na.rm = TRUE),
                                          mean = mean(conf_wins_pct, na.rm = TRUE)
                                          )
reg_wins_pct_summary <- WMM_tidy_data |> 
                                summarize(min = min(reg_wins_pct, na.rm = TRUE),
                                          max = max(reg_wins_pct, na.rm = TRUE),
                                          mean = mean(reg_wins_pct, na.rm = TRUE)  
                                          )
WMM_summary_data <- bind_rows(Year_summary, # I used bind_rows() to combine the 4 tibbles  (one for each numerical variable) that I'd created into 1 tibble with the min, max, and mean for each variable. 
                              Seed_summary,
                              conf_wins_pct_summary,
                              reg_wins_pct_summary) |>
                                mutate(variable_names = c("year", # I needed to add the variable names to the front, so I used mutate and c to create a new column called "variable_names" with all the variable names. 
                                                          "seed",
                                                          "conf_wins_pct",
                                                          "reg_wins_pct")) |> 
                                select(variable_names, # the variables_names column was added to the end of the table, so I used select() to push it to the front of the table
                                                   min,
                                                   max,
                                                    mean)

WMM_summary_data
```

This boxplot is relevant for my analysis because it contrasts the seeds (a team's preliminary tournament ranking) with where the team ended up finishing in the tournament. From the data, its clear that the champion almost always starts with a high initial seed (the lowest initial seed of any champion ever is 3rd seed), and those who get removed early from the tournament usually start at worse inital seeds. This means that starting seed is likely a good indicator of a team's tournament finish. 

```r
boxplot1 <- WMM_tidy_data |> mutate(tourney_finish = fct_recode(tourney_finish, # I renamed the tournament finish classes to be more descriptive https://www.r-bloggers.com/2024/03/title-how-to-rename-factor-levels-in-r-with-examples/
                                               "Second Round Loss" = "srl",
                                               "First Round Loss" = "frl",
                                               "Top 16 Loss" = "t16" ,
                                               "Top 8 Loss" = "t8",
                                               "Top 4 loss" = "t4",
                                               "Top 2 loss" = "t2")) |>
                            ggplot(aes(y = tourney_finish, x = seed)) + 
                                geom_boxplot() +   # I decided to use a boxplot because tourney_finish is a categorical variable
                                labs(title = "Team Initial Seed versus Tournament Finish", x = "Initial Seed", y = "Tournament Finish")
boxplot1
```

This analysis shows the regional win percentage for each tournament finish class. On average, WMM tournament champions won close to 90% of their regional games, while, on average, teams that lost in the first round won less than 75% of their regional games. The pattern that can be seen here is that, on average, teams that performed worse in regionals are likely to do worse in the tournament, while teams that did better in regionals are much more likely to perform better in the tournament. This makes regional win percentages a likely good predictor for tournament finish.

```r
barplot1 <- WMM_tidy_data |> 
 mutate(tourney_finish = fct_recode(tourney_finish, # I renamed the tournament finish classes to be more descriptive https://www.r-bloggers.com/2024/03/title-how-to-rename-factor-levels-in-r-with-examples/
                                               "Second Round Loss" = "srl",
                                               "First Round Loss" = "frl",
                                               "Top 16 Loss" = "t16" ,
                                               "Top 8 Loss" = "t8",
                                               "Top 4 loss" = "t4",
                                               "Top 2 loss" = "t2")) |>
            group_by(tourney_finish) |> 
            summarize(avg_reg_wins_pct = mean(reg_wins_pct)) |>
            ggplot(aes(x = avg_reg_wins_pct, y = fct_reorder(tourney_finish, avg_reg_wins_pct))) + # I used fct_reorder to order the tournament finish classes by average regional games win %
            geom_bar(stat = "identity") +
            labs(x = "Average Regional Games Win Percentage", y = "Tournament Finish", title = "Average Regional Games Win Percentage By Tournament \n Finish")
barplot1
```

My 2nd bar plot shows the conference games win percentage by tournament finish. The pattern that can be seen here is that, in general, teams that win a greater percentage of their conference games do better in the WMM tournament, while teams that perform worse in their conference games usually end up losing early in the WMM tournament. This means that conference win percentages may be a good predictor of tournament finish. 

```r
barplot2 <- WMM_tidy_data |> 
 mutate(tourney_finish = fct_recode(tourney_finish, # I renamed the tournament finish classes to be more descriptive https://www.r-bloggers.com/2024/03/title-how-to-rename-factor-levels-in-r-with-examples/
                                               "Second Round Loss (Round of 32)" = "srl",
                                               "First Round Loss (Round of 64)" = "frl",
                                               "Top 16 Loss" = "t16" ,
                                               "Top 8 Loss" = "t8",
                                               "Top 4 loss" = "t4",
                                               "Top 2 loss" = "t2")) |>
            group_by(tourney_finish) |> 
            summarize(avg_conf_wins_pct = mean(conf_wins_pct)) |>
            ggplot(aes(x = avg_conf_wins_pct, y = fct_reorder(tourney_finish, avg_conf_wins_pct))) + # I used fct_reorder to order the tournament finish classes by average conference games win %
            geom_bar(stat = "identity") +
            labs(x = "Average Conference Games Win Percentage", y = "Tournament Finish", title = "Average Conference Games Win Percentage By \n Tournament Finish")
barplot2
```
